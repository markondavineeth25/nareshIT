### 9. Concatenation of Array

The arrays can be concatenated only if they have the same shape, except in the dimension corresponding to the axis of concatenation.

**Concatenate 1D array**

```python
# concatenate two 1D arrays
# production on day 1
production_day_1 = np.array([11, 22, 13])

# production on day 2
production_day_2 = np.array([23, 22, 12])

```

```python
prod_data_2_days = np.concatenate([production_day_1, production_day_2])
prod_data_2_days
```

You can also concatenate more than two arrays at once.

```python
# production on day 3
production_day_3 = np.array([23,45])
```

```python
np.concatenate([prod_data_2_days, production_day_3])
```

```python
np.concatenate([production_day_1, production_day_2, production_day_3])
```

**Concatenate 2D array**

```python
# create a 2D array
scores_player_1 = np.array([[1, 2, 3],
                 [4, 5, 6]])
print(scores_player_1)
print(scores_player_1.shape)
print(scores_player_1.ndim)
```

```python
# create a 2D array
scores_player_2 = np.array([[5, 6, 3],
                 [4, 9, 6]])
print(scores_player_2)
print(scores_player_2.shape)
print(scores_player_2.ndim)
```

**concatenate along the first axis**

```python
np.concatenate((scores_player_1, scores_player_2), axis = 0)
```

**concatenate along the second axis**

```python
np.concatenate((scores_player_1, scores_player_2), axis = 1)
```

### 13. Pandas DataFrames

 A DataFrame is a tabular representation of data containing an ordered collection of columns, each of which can be a different type (such as numeric, string, boolean). <br><br>
                        The DataFrame has both row and column index; it can be thought of as a dict of Series all sharing the same index. In a data frame, the data is stored as one or more two-dimensional blocks rather than a list, dict, or some other collection of one-dimensional arrays. <br><br>
                        While a DataFrame is physically two-dimensional, it can be used to represent higher dimensional data in a tabular format using hierarchical indexing.
