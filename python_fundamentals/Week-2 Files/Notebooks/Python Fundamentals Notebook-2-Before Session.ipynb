#### Comprehensive Definition of Data Structures in Python

Data structures in Python are specialized ways of storing, organizing, and managing data so that it can be accessed, modified, and processed efficiently.

They define how data is arranged in memory and how various operations—such as searching, inserting, deleting, and sorting—can be performed on that data with optimal performance.

In simple words, a data structure acts like a container that holds related data items together, allowing programmers to handle complex data relationships easily.

### Accessing Elements in a List
You can access the list items by referring to the index number

**Indexing:**<br>
Use the index operator [ ]. Each element in a list has an assigned index value. Note that the indexing in python starts from 0 and every index must be an integer

```python
# defining a list
course = ['data science', 'machine learning', 'python', 'html', 'big data' ]
```

```python
course = ['data science', 'machine learning', 'python', 'html', 'big data' ]
print(course[0])
print(course[-1])
```

```python
# multi level indexing
course[1][0]
```

**Negative Indexing**:<br>
Negative indexing means position count starts from the end of the list, -1 refers to the last item, -2 refers to the second last item

```python
course[-1]
```

**Length of List:**<br>
Use len() function to find total number of elements in a list

```python
len(course)
```

```python
len(course[1])
```

### Slicing of a List
List uses the slicing operator : (colon) to extract part of the sequence

### Syntax: sequence[start : end : step]

start → the index where the slice begins (included)<br>
end → the index where the slice stops (excluded)<br>
step → how many items to skip (optional)<br>

If all three 'start', 'stop', and 'step' are provided, it generates portion of sequence beginning from 'start' till 'stop' with increment of 'step'

```python
course
```

```python
course[::]
```

```python
course[:3:1]
```

```python
course[::2]
```

###### list reverse

```python
course[::-1]
```
