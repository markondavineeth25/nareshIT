#### Comprehensive Definition of Data Structures in Python

Data structures in Python are specialized ways of storing, organizing, and managing data so that it can be accessed, modified, and processed efficiently.

They define how data is arranged in memory and how various operations—such as searching, inserting, deleting, and sorting—can be performed on that data with optimal performance.

In simple words, a data structure acts like a container that holds related data items together, allowing programmers to handle complex data relationships easily.

### Create Dictionary
Separate keys from values with a colon(:), and a pair from another by a comma(,). Put it all in curly braces

```python
### Membership Operators in Dictionary
# Create a dictionary
student_info = {"name": "Alice", "course": "Data Science", "ID": 101}

# Check if a key exists using 'in'
print("Is 'name' a key in student_info?", "name" in student_info)

# Check if a key does not exist using 'not in'
print("Is 'age' not in student_info?", "age" not in student_info)

# Note: Membership operators check keys by default in dictionaries
print("Is 'Data Science' in student_info?", 'Data Science' in student_info)
```

### Accessing Items
Use the key name inside the square brackets to access the elements from a dictionary

```python
### Accessing items using a key
# Define a sample dictionary
scores = {"Mia": 85, "John": 92, "Leo": 78}

# Get the value for key "Mia"
mia_score = scores["Mia"]
print(f"The score for Mia is: {mia_score}", mia_score)
```

We can use get method to get the value of "Mia" key

```python
# Using the get() method to access a value
# Scores dictionary was defined as: {'Mia': 85, 'John': 92, 'Leo': 78}

mia_score_get = scores.get('Mia')
print(f"Score for Mia (using get): {mia_score_get}")

# Example of handling a missing key with a default value
sara_score = scores.get('Sara', 'Not Found')
print(f"Score for Sara: {sara_score}")
```

#### Clarification on `set.pop()`

As mentioned, `set.pop()` removes an **arbitrary** element from the set. This means you cannot predict exactly which element will be removed. It's not truly 'random' in the sense of a uniform distribution, but rather relies on the internal implementation details of how sets are stored (which can change across Python versions).

If you need to remove a *specific* element, use `set.remove(element)` or `set.discard(element)`.

```python
# Let's demonstrate set.pop() behavior.
# Note: The output might vary based on Python version and execution environment.

my_colorful_set = {'red', 'green', 'blue', 'yellow', 'purple'}
print(f"Original set: {my_colorful_set}")

popped_item_1 = my_colorful_set.pop()
print(f"Popped item 1: {popped_item_1}")
print(f"Set after first pop: {my_colorful_set}")

popped_item_2 = my_colorful_set.pop()
print(f"Popped item 2: {popped_item_2}")
print(f"Set after second pop: {my_colorful_set}")
```

### Set Operations

Disjoint set: if the two sets have no common elements

```python
print (set([11, 12, 22, 33, 4]).isdisjoint(set([4, 1])))
```

**Checking for subset**<br>
Use issubset() method to check whether all elements of a set are contained in another set

```python
# check whether one set 'a' is a subset of the 'b'
a = set([1, 3, 4, 5])
b = set([1, 3, 4, 5, 6])
print('Is a subset of b?', a.issubset(b))
```

**Set Union**<br>
Use union() method to compute the union of two or more sets

```python
a = set([1, 2, 3])
b = set([3, 4, 5, 6])
print (a.union(b))
```

**Set Intersection**<br>
Take intersection to find the common elements

```python
a = set([12,34,56,78])
b = set([12,45,67,89,78])
c = set([12,34,56,78,56])
print (a & b)
print (a & b & c)
```

**Set Difference**<br>
It returns a new set containing all items from the first set that are not in the other set

```python
a = set([23,45,56,77,89])
b = set([23,45,54,33,23])
print(a - b)
```
