##### Write a Python program using nested if statements to check if the student is eligible to write the exam.

A student is allowed to write the exam only if both conditions are met:<br>
Age is 18 or above<br>
Attendance is 75% or more<br>


```python
age = int(input('Enter Age :'))
attendance = int(input('Enter Attendance :'))
if age >= 18:                     # Outer if
    if attendance >= 75:          # Inner if (nested)
        print("Eligible")
    else:
        print("Not eligible due to low attendance")
else:
    print("Not eligible due to age")
```

```python
# For loop to print even numbers from 1 to 20
print("Even numbers from 1 to 20:")
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

##### Write a Python program that creates an empty list and uses a for loop to append all even numbers between 1 and 20 to the list. Finally, print the list.

```python
# Create an empty list
even_numbers = []

# Use a for loop to iterate from 1 to 20
for i in range(1, 21):
    # Check if the number is even
    if i % 2 == 0:
        # Append even numbers to the list
        even_numbers.append(i)

# Print the final list
print("Even numbers between 1 and 20:", even_numbers)
```

##### Write a Python program that counts how many even numbers are in the given list.

```python
numbers = [1, 4, 7, 10, 12, 15, 18]
count = 0

for num in numbers:
    if num % 2 == 0:
        count += 1

print("Total even numbers:", count)
```

### Built-in Functions


#### The Python interpreter has functions instantly available for use, called <i>built-in functions</i>. For example, the print() function prints the given object.

##### Print() function: Used to display output on the screen.

```python
print("Hello, Python!")
# Output: Hello, Python!
```

##### len() function: Returns the number of items in a string, list, tuple, or any collection.

```python
name = "Rakesh"
print(len(name))
# Output: 6
```

##### type() function: Tells you the data type of a variable.

```python
x = 10
print(type(x))
# Output: <class 'int'>
```

##### sum() function: Adds up all numbers.

```python
numbers = [5, 10, 15]
print(sum(numbers))
# Output: 30
```

##### abs() function: Gives the absolute (positive) value of a number.

```python
print(abs(-25))
# Output: 25
```

##### min() & max() functions: Return the largest and smallest values in a collection.

```python
marks = [65, 89, 72, 90]
print(max(marks))  # Output: 90
print(min(marks))  # Output: 65
```

### User Defined Functions

#### There are situations where built-in functions are not available in a library. In such cases, we do the computations by using formulas/concepts

#### Argument and Parameter

Parameters and arguments are subtly different. Let us first define them.<br>
1. Parameter: A value which is passed when defining a function<br>
2. Argument: A value that is passed when invoking a function<br>
                    

Parameter → A placeholder or variable inside a function definition.<br>
It tells the function what kind of information it will receive.<br>

Argument → The actual value you pass to the function when you call it.<br>
Think of it like a box and the gift you put inside<br>

The parameter is the empty box (it expects something).<br>
The argument is the gift you actually put inside the box.

```python
# Function definition
def greet(name):        # 'name' is the PARAMETER
    print("Hello, " + name + "!")

# Function call
greet("Rakesh")         # "Rakesh" is the ARGUMENT
```

#### User defined functions without arguments

**1. We shall write a user defined function without argument to find the twice of a number.**

```python
def twice():
    a = int(input('enter a number: '))
    print(a*2)
```

**2. To find area of a circle**

```python
def area_circle():
    r = int(input('Enter the radius of circle: '))
    area = 3.14 * r**2 
    print("The area of circle with radius " + str(r)  +" is " + str(area) + " sq. units")
```

```python
area_circle()
```

**2. A user defined function with argument to Check if given number Even or Odd?**

```python
def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

print("7 is", check_even_odd(7))
print("10 is", check_even_odd(10))
```

##### Basic Syntax

[expression for item in iterable if condition]


###### Normal way to print numbers from 1 to 10 using for loop. 

```python
for i in range(1,10):
    print(i,end=' ')
```

###### Using List Comprehension 

```python
num=[i for i in range(1,10)]
num
```

###### Normal Way of finding squares of numbers (using a loop)

```python
squares = []
for i in range(1, 6):
    squares.append(i ** 2)

print(squares)
```

###### Using List Comprehensions

```python
# Finding Squares of numbers from 1 to 6.
squares = [i ** 2 for i in range(1, 6)]
print(squares)
```

```python
# Finding even numbers in a list 
even = [i for i in range(1,11) if i%2==0]
even
```

### 4.1 Function map()

**1. A test was conducted in a class, the list of all the scores is given. Tha scores are out of 45. Obtain the percentage of each score using lambda function.**
             
             marks = [ 20, 23, 34, 45, 33, 10, 43, 23, 27, 34, 37, 44, 42, 43 ]


```python
marks = [ 20, 23, 34, 45, 33, 10, 43, 23, 27, 34, 37, 44, 42, 43 ]
```

```python
percentages = list(map(lambda x: (x / 45) * 100, marks))
print(percentages,end=' ')
```

**2. Create two lists and add corresponding elements of each list**

```python
# Two sample lists
list1 = [10, 20, 30, 40, 50]
list2 = [1, 2, 3, 4, 5]

# Using map() with lambda to add corresponding elements
result = list(map(lambda x, y: x + y, list1, list2))

print(result)
```

<a id='filter'></a>
### 4.2 Function filter()

**1. To obtain the odd numbers from a list**

```python
# Given list
numbers = [10, 15, 22, 33, 40, 55, 60, 77, 80, 91]

# Use filter() to get only odd numbers
odd_numbers = list(filter(lambda x: x % 2 != 0, numbers))

# Print result
print("Odd numbers are:", odd_numbers)
```

### 4.3 Function reduce()

```python
from functools import reduce
```

#### Find the Sum of All Numbers

numbers = [1, 2, 3, 4, 5]

```python
numbers = [1, 2, 3, 4, 5]
result = reduce(lambda x, y: x + y, numbers)
print("Sum of all numbers:", result)
```
