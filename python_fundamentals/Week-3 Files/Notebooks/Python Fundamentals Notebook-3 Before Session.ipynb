Let’s see an example on the if…else statement.

```python
num = 10
if num % 2 == 0:
    print(f'{num} is an even number.')
else:
    print(f'{num} is an odd number.')
```

```python
num = 11
if num % 2 == 0:
    print(f'{num} is an even no.')
else:
    print(f'{num} is odd no.')
```

```python
age = 10 # You can change this value to test different ages
if 9 <= age <= 12:
    print(f'The person is {age} years old, which is between 9 and 12.')
else:
    print(f'The person is {age} years old, which is not between 9 and 12.')
```

#### Using the `and` keyword for multiple conditions
The `and` keyword is a logical operator that allows you to combine two or more conditional statements. The entire condition evaluates to `True` only if **all** individual conditions connected by `and` are `True`.

```python
num_to_check = 25

# Check if num_to_check is greater than 20 AND less than 30
if num_to_check > 20 and num_to_check < 30:
    print(f'{num_to_check} is between 20 and 30.')
else:
    print(f'{num_to_check} is not between 20 and 30.')

num_to_check = 15
if num_to_check > 20 and num_to_check < 30:
    print(f'{num_to_check} is between 20 and 30.')
else:
    print(f'{num_to_check} is not between 20 and 30.')
```

```python
num = 12

if num <10 or num > 15:
    print(f'{num} is not between 10 and 15')
else:
    print(f'{num} is between 10 and 15')
```
